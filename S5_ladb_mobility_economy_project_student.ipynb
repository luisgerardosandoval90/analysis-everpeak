### 3.1 Extraer columna año y filtrar 2024

**🎯Objetivo**
Identificar el año de cada registro y mantener solo los registros del 2024.

**Intrucciones**

- Como el DataFrame `traffic` no tiene una columna de año, utiliza el atributo `.dt.year` sobre su columna de fecha para crear una nueva columna llamada `year`.
- Filtra las filas donde el año sea **2024**.
- Utiliza `.copy()` para crear dos nuevos DataFrames (`traffic_2024` y `eco_2024`) para evitar modificar el dataset original.

```python

# 1. Asegurar la conversión a datetime (por si venía como texto)
traffic['update_time_utc'] = pd.to_datetime(traffic['update_time_utc'], errors='coerce')

# 2. Ahora sí extraemos el año de las fechas en update_time_utc
traffic['year'] = traffic['update_time_utc'].dt.year

# Verificar cambio
traffic.head(3)

```

## 🧩Paso 4: Analizar y resumir datos de movilidad

Como el dataset de tráfico contiene **múltiples registros por ciudad**. En esta parte, calcularás los promedios anuales por ciudad para simplificar el análisis y obtener una visión más clara de las tendencias generales.

### 4.1 Calcular promedios de tráfico por ciudad

**🎯Objetivo:**
Obtener una vista consolidada del tráfico promedio por ciudad y año, para analizar patrones generales sin depender de datos diarios.

**Instrucciones**

- Agrupa los datos por `city`, `country` y `year`.
- Calcula el promedio **solo de las métricas de tráfico más relevantes**: como `jams_delay`, `traffic_index_live`, `jams_length_kms`, `jams_count`, `mins_delay`, y tiempos de viaje (`travel_time_live_per_10kms_mins` y `travel_time_hist_per_10kms_mins`).
- Guarda el resultado como `traffic_city_year_2024`, mantén las columnas como variables (no índices).


<details>
<summary>Haz clic para ver la pista</summary>
Usa ".agg()" para aplicar funciones de promedio. Al final, reinicia el índice para mantener las columnas de la agrupación como variables (no índices).

```python

# Calcular los promedios de tráfico por ciudad, país y año
traffic_city_year_2024 = (traffic_2024
    .groupby(['city', 'country', 'year'], as_index=False)[
        [
            'jams_delay',
            'traffic_index_live',
            'jams_length_in_kms',
            'jams_count',
            'mins_delay',
            'travel_time_live_per_10kms_mins',
            'travel_time_historic_per_10kms_mins'
        ]
    ]
    .mean()
)

# Mostrar resultado
traffic_city_year_2024.head()
```

## 🧩Paso 5: Unir movilidad y economía

Combinar datasets te permite analizar cómo se relacionan los indicadores económicos con los de movilidad.

### 5.1 Unir tráfico (tabla principal) con indicadores económicos

**🎯Objetivo:**
Combinar la información de tráfico y economía en un solo DataFrame para analizar cómo las condiciones económicas se relacionan con la movilidad urbana.

**Instrucciones**
- Selecciona solo las **columnas relevantes** de cada dataset (por ejemplo, variables clave de tráfico y de economía).
- Usa `.copy()` al crear subconjuntos para evitar modificar el dataset original.
- Une ambos DataFrames y define como **claves de unión** a `city` y `year`.
- Mantén solo las ciudades y años presentes en ambos datasets.
- Guarda el resultado en una nueva variable llamada `merged` y muestra las primeras 5 filas.


<details>
<summary>Haz clic para ver la pista</summary>
Aplica una unión de tipo "inner" para mantener las ciudades y años presentes en ambos datasets.

```python
# Seleccionar columnas clave de tráfico y economía
left_cols = [
    'city', 'country', 'year', 'jams_delay', 'traffic_index_live',
    'jams_length_in_kms', 'jams_count', 'mins_delay',
    'travel_time_live_per_10kms_mins', 'travel_time_historic_per_10kms_mins'
]

right_cols = ['city', 'year', 'city_gdp_capita', 'unemployment_pct', 'pm25_μg_m³', 'population']

# Usar .copy() para crear los dos nuevos datasets reducidos
traffic_2024_small = traffic_city_year_2024[left_cols].copy()
eco_2024_small = eco_2024[right_cols].copy()

# Unir datasets mediante un inner join usando 'city' y 'year' como claves
merged = pd.merge(
    traffic_2024_small,
    eco_2024_small,
    on=['city', 'year'],
    how='inner'
)

# Mostrar las primeras 5 filas
merged.head()
```

# 🧾 Resumen ejecutivo

**Contexto & objetivo:**  
Este análisis evalúa la relación entre los indicadores de movilidad urbana (principalmente el retraso vehicular `jams_delay`) y la productividad económica (PIB per cápita `city_gdp_capita`). El propósito principal es determinar si un mayor nivel socioeconómico mitiga la congestión o si existen factores estructurales predominantes, proporcionando evidencia cuantitativa para respaldar la toma de decisiones e inversión en infraestructura de transporte público.

**Cobertura de datos:**  
El dataset consolidado incluye registros del año **2024** para **15 ciudades principales** (centradas prioritariamente en América Latina y economías emergentes), integrando variables de tráfico, PIB per cápita, desempleo, nivel de contaminación (PM2.5) y población total.

**Metodología (alto nivel):**  
- **Limpieza y estandarización:** Conversión de registros temporales a tipo `datetime`, eliminación de caracteres especiales en valores numéricos y estandarización del esquema de nombres de columnas a `snake_case`.
- **Agregación e integración:** Filtrado del año 2024, cálculo de promedios anuales por ciudad mediante `.groupby()` y posterior cruce de datos mediante `pd.merge(how='inner')` utilizando `city` y `year` como llaves de unión.
- **Validación visual y estadística:** Evaluación de distribuciones con boxplots e histogramas, análisis visual comparativo con gráficos de barras de doble eje y cálculo de matrices de correlación de Pearson.

**Hallazgos iniciales:**  
- **Débil relación lineal entre PIB y Congestión ($r = 0.2830$):** Los datos demuestran que un mayor PIB per cápita no garantiza menores tiempos de congestión. Por ejemplo, Montevideo registra el PIB per cápita más alto del grupo ($26,176 USD) con una congestión mínima (50.20 min), mientras que Mexico City combina un PIB alto ($21,111 USD) con el nivel de congestión más elevado del dataset (2,833.06 min).
-
- **Predominio del volumen poblacional ($r = 0.8785$):** La congestión vehicular presenta una correlación fuertemente positiva con el tamaño de la población, evidenciando que la densidad y la escala urbana influyen más en el tráfico que el nivel de ingreso per cápita por sí solo.
- **Valores atípicos y severidad:** Mexico City (2,833.06 min) y Sao Paulo (1,729.19 min) representan los valores atípicos absolutos en retraso vehicular.

**Recomendaciones:**  
- **Prioridad de inversión (Bogotá y Lima):** Se identifica a **Bogotá** (`jams_delay` = 1,141.55 min; PIB per cápita = $11,442 USD) y **Lima** (`jams_delay` = 1,052.29 min; PIB per cápita = $13,472 USD) como las ciudades prioritarias para la asignación de recursos e infraestructura de transporte masivo. Aunque Mexico City encabeza los retrasos absolutos, Bogotá y Lima presentan índices de congestión extremadamente elevados en proporción a su población (~11.3 millones de habitantes) combinados con ingresos per cápita moderados, lo que genera un impacto negativo severo en la productividad y movilidad de sus habitantes.
- **Acciones estratégicas:** Priorizar la expansión de redes de transporte público de alta capacidad (sistemas de metro y corredores BRT) en lugar de infraestructura orientada al vehículo particular, así como incorporar variables de densidad de red vial en futuros modelos analíticos.
